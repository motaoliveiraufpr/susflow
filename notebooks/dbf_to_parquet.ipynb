```python
from pathlib import Path
from dbfread import DBF
import pandas as pd
```

```python
IN_COLAB = False
try:
    from google.colab import drive
    IN_COLAB = True
except Exception:
    IN_COLAB = False

if IN_COLAB:
    drive.mount('/content/drive')
    DATA_DIR = Path('/content/drive/MyDrive/Projeto Indicadores/data')
else:
    DATA_DIR = Path(r'C:\Users\Avell\Desktop\Projeto\data')

CSV_DIR_SIH = DATA_DIR / 'raw' / 'SIH'
PARQUET_DIR_SIH = DATA_DIR / 'parquet' / 'SIH'
```

```python
tabela = DBF(CSV_DIR_SIH / 'RDPR2512.dbf', encoding='latin1', load=True)
```

```python
df = pd.DataFrame(iter(tabela))
```

```python
df.to_parquet(PARQUET_DIR_SIH / 'RDPR2512.parquet', index=False)
```
