```python
from pathlib import Path
import osmnx as ox
```

```python
IN_COLAB = False
try:
    from google.colab import drive
    IN_COLAB = True
except Exception:
    IN_COLAB = False

if IN_COLAB:
    drive.mount('/content/drive')
    DATA_DIR = Path('/content/drive/MyDrive/Projeto Indicadores/data')
else:
    DATA_DIR = Path(r'C:\Users\Avell\Desktop\Projeto\data')

GRAPHML_DIR = DATA_DIR / 'graphml'
GRAPHML_FILE = GRAPHML_DIR / 'road_network_2hr.graphml'
```

```python
ox.settings.use_cache = True
```

```python
west = -50.1
south = -26.3
east = -47.9
north = -24.5
bbox = (west, south, east, north)
```

```python
G_road = ox.graph_from_bbox(
    bbox=bbox,
    network_type="drive"
)
```

```python
ox.save_graphml(G_road, GRAPHML_FILE)
```
